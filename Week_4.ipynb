A Series is a one-dimensional labeled array capable of holding any data type.

#Importing, Cleaning, and Transforming Biological Data

**a) Importing Data**

Pandas makes it easy to import data from various file formats, including CSV, Excel, and more.

````python
# Example: Importing data from a CSV file and Check for Missing Values
import pandas as pd

# Step 1: Load the CSV file into a DataFrame
df = pd.read_csv('gene_expression.csv')

# Step 2: Check if there are any missing values in the DataFrame
if df.isnull().values.any():
    print("Missing values detected!")

    # Step 3: Count the number of missing values in each column
    missing_values_count = df.isnull().sum()
    print("Number of missing values in each column:\n", missing_values_count)
else:
    print("No missing values detected.")

# Step 4: Optionally, display the rows with missing values
rows_with_missing = df[df.isnull().any(axis=1)]
print("\nRows with missing values:\n", rows_with_missing)
````

````python
print(df.head())  # Display the first few rows of the DataFrame
````

**Explanation:**

- read_csv(): Reads data from a CSV file into a DataFrame. The head() function displays the first few rows, giving a quick overview of the data.

**b) Cleaning Data**

Cleaning data is often a necessary step in data analysis. This might involve handling missing values, removing duplicates, or converting data types.

````python
# Example: Handling missing values
df.fillna(0, inplace=True)  # Replace missing values with 0
print(df.isnull().sum())  # Check for any remaining missing values

# Example: Removing duplicates
df.drop_duplicates(inplace=True)
print(df)
````

**Explanation:**

- fillna(): Replaces missing values with a specified value.
- drop_duplicates(): Removes duplicate rows from the DataFrame.

**c) Transforming Data**

Transforming data involves changing the structure or values in the DataFrame to make it more suitable for analysis.

Ensure that the Expression_Level Column is Numeric
First, you need to ensure that the Expression_Level column is converted to a numeric data type. If there are any non-numeric characters or missing values, you can use pd.to_numeric() to convert the column and handle errors appropriately.

````python
import pandas as pd

# Example DataFrame based on your provided data
data = {
    'Gene_symbol': ['Aaas', 'Aacs', 'Aaed1', 'Aagab', 'Aak1'],
    'NAME': [
        'achalasia; adrenocortical insufficiency; alacrimia',
        'acetoacetyl-CoA synthetase',
        'AhpC/TSA antioxidant enzyme domain containing 1',
        'alpha- and gamma-adaptin binding protein',
        'AP2 associated kinase 1'
    ],
    'Expression_Level': ['1.817065654', '1.309325862', '-1.326889651', '1.271087151', '-1.030009115'],
    'PValue': ['3.67341681694533e-14', '0.000143671', '0.007682286', '0.000957723', '0.700646883'],
    'FDR': ['1.20044643552425e-12', '0.000764351', '0.023586981', '0.004004751', '0.79706867']
}

df = pd.DataFrame(data)

# Convert the 'Expression_Level' column to numeric, coercing errors to NaN
df['Expression_Level'] = pd.to_numeric(df['Expression_Level'], errors='coerce')

# Check if the conversion was successful
print(df['Expression_Level'])

# Now you can safely add the normalized column
df['Normalized_Expression'] = df['Expression_Level'] / df['Expression_Level'].max()
print(df)

# Check data types
print(df.dtypes)
````

**Explanation:**

1. Convert to Numeric with pd.to_numeric():

- The pd.to_numeric() function attempts to convert the Expression_Level column to a numeric type.

- The errors='coerce' argument will turn any non-numeric values (like strings or special characters) into NaN, ensuring that the column becomes fully numeric.

2. Handling Missing or Invalid Data:

- If there are any non-numeric values that cannot be converted, they will become NaN. You should decide how to handle these NaN values (e.g., filling them with a default value or dropping them).

3. Adding the Normalized Column:

- Once the Expression_Level column is confirmed to be numeric, you can safely compute the Normalized_Expression by dividing each value by the maximum value in that column.

4. Check Data Types:

- The print(df.dtypes) command will show you the data types of each column, ensuring that the conversion was successful.

**Result:**
After running this code, the Expression_Level column should be numeric, and you should be able to add the Normalized_Expression column without encountering any errors.

#Data Analysis Techniques

**a) Filtering Data**

Filtering allows you to select specific rows in a DataFrame based on a condition.

````python
import pandas as pd

# Example data for gene expression
data = {
    'Gene': ['BRCA1', 'TP53', 'EGFR', 'MYC', 'BRCA1', 'TP53'],
    'Expression_Level': [12.5, 8.3, 14.7, 7.1, 10.8, 9.5],
    'Organism': ['Human', 'Human', 'Human', 'Mouse', 'Mouse', 'Mouse'],
    'PValue': [0.001, 0.03, 0.002, 0.045, 0.005, 0.015],
    'FDR': [0.01, 0.04, 0.02, 0.05, 0.02, 0.03],
    'Sample1': [23.7, 25.6, 24.3, 29.5, 25.2, 28.1],
    'Sample2': [22.9, 28.2, 19.6, 30.3, 24.4, 26.5],
    'Sample3': [22.8, 20.1, 26.1, 30.8, 24.5, 25.7],
}

# Convert to a DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv('gene_expression_1.csv', index=False)
````

````python
# Example: Filtering for genes with high expression levels
high_expression = df[df['Expression_Level'] > 10]
print(high_expression)
````

**Explanation:**

- Filtering: Pandas makes it easy to filter data based on conditions, helping you focus on specific subsets of your data.

**b) Grouping Data**

Grouping data is useful for aggregating data based on specific categories.

````python
# Example: Grouping by organism and calculating the mean expression level
grouped = df.groupby('Organism')['Expression_Level'].mean()
print(grouped)
````

**Explanation:**

- Grouping: The groupby() function allows you to group data by one or more columns and perform aggregate operations, such as calculating the mean.

**c) Aggregation**

Aggregation refers to applying summary statistics (e.g., sum, mean, count) to grouped data.

````python
# Example: Aggregating data to get mean and max expression levels
aggregated = df.groupby('Organism').agg({'Expression_Level': ['mean', 'max']})
print(aggregated)
````

**Explanation:**

- Aggregation: The agg() function allows you to apply multiple aggregation operations on grouped data, providing deeper insights.

#Introduction to Python's Built-In Libraries


**1. The collections Module**

**Overview:**

The collections module provides specialized container datatypes beyond Python's built-in types like lists, dictionaries, and tuples. These structures can be particularly useful in bioinformatics for managing and manipulating data efficiently.

**a) defaultdict**

What It Is: defaultdict is a dictionary that automatically assigns a default value when a missing key is first accessed. For example, defaultdict(int) initializes missing keys to 0, making it useful for counting occurrences without checking whether a key already exists.

````python
from collections import defaultdict

# Example: Counting nucleotide occurrences in a DNA sequence
sequence = "ATGCGCATTAGCGATCGATC"
nucleotide_count = defaultdict(int)

for nucleotide in sequence:
    nucleotide_count[nucleotide] += 1

print(dict(nucleotide_count))
````

**Explanation:**

The key is defaultdict(int).
- int provides a default value of 0.
- When a nucleotide such as "A" is encountered for the first time, defaultdict automatically creates "A": 0.
- The statement += 1 then increases its count to 1.
- Each subsequent occurrence increases the count by another 1.

**b) Counter**

What It Is: A dictionary subclass specifically designed for counting hashable objects, often used to count elements in a collection.

````python
from collections import Counter

# Example: Counting nucleotides in a DNA sequence
sequence = "ATGCGCATTAGCGATCGATC"
nucleotide_count = Counter(sequence)

print(nucleotide_count)
````

**Explanation:**

- Counter: Quickly counts the occurrences of each element in the sequence and stores the results in a dictionary-like structure.

**c) namedtuple**

What It Is: A factory function for creating tuple subclasses with named fields, allowing for more readable and self-documenting code.

````python
from collections import namedtuple

# Example: Creating a named tuple for a gene
Gene = namedtuple('Gene', 'name sequence expression_level')
gene1 = Gene(name="BRCA1", sequence="ATGCGC", expression_level=12.5)

print(gene1)
print(gene1.name)
print(gene1.sequence)
````

**Explanation:**

- namedtuple: Provides a lightweight and efficient way to store related data together, like gene information, while allowing access to fields by name instead of by index.

**2. The itertools Module**

**Overview:**

The itertools module provides a collection of tools for efficient looping and iteration. These tools are invaluable for tasks like generating combinations, permutations, or iterating over multiple sequences simultaneously, which are common in bioinformatics.

**a) product**

What It Is: Computes the Cartesian product of input iterables, useful for generating all possible combinations of elements.

````python
from itertools import product

# Example: Generating all possible dinucleotides
bases = ['A', 'T', 'G', 'C']
dinucleotides = list(product(bases, repeat=2))

print(dinucleotides)
````

**Explanation:**

- product: Generates all possible pairs of nucleotides, which is useful for tasks like enumerating all possible k-mers in sequence analysis.

**b) combinations**

What It Is: Generates all possible combinations of a specified length from the input iterable.



````python
from itertools import combinations

# Example: Finding all pairs of nucleotides
pairs = list(combinations('ATGC', 2))

print(pairs)
````

**Explanation:**

- combinations: Useful for generating pairs (or larger sets) of nucleotides, genes, or other biological entities.

**c) cycle and islice**

What They Are: cycle repeats an iterable indefinitely, and islice slices an iterator, allowing you to take only the first n elements.

````python
from itertools import cycle, islice

# Example: Cycling through nucleotides
cyclic_bases = cycle('ATGC')
first_ten = list(islice(cyclic_bases, 20))

print(first_ten)
````

**Explanation:**

- cycle and islice: These functions are helpful when you need to repeat a sequence multiple times or when working with large datasets where you want to limit the number of items processed at a time.

- Library: itertools
- Module: itertools
- Functions: cycle and islice

In this case, Python's standard library includes itertools as a standalone module that provides various iterator-building functions, so itertools serves as both the library and module.

**Practical Exercise: Applying Python's Built-In Libraries**

Task: Write a Python script that:

- Uses Counter from the collections module to count the frequency of each nucleotide in a given DNA sequence.
- Generates all possible trinucleotides using product from the itertools module.
- Lists all .fasta files in the current directory using the os module.
- Accepts a DNA sequence as a command-line argument using the sys module.

````python
import os
import sys
from collections import Counter
from itertools import product

# Part 1: Counting nucleotide frequency using Counter
sequence = "ATGCGCATTAGCGATCGATC"
nucleotide_count = Counter(sequence)
print("Nucleotide frequency:", nucleotide_count)

# Part 2: Generating all possible trinucleotides
trinucleotides = list(product('ATGC', repeat=3))
print("All possible trinucleotides:", trinucleotides)

# Part 3: Listing all .fasta files in the current directory
fasta_files = [f for f in os.listdir('.') if f.endswith('.fasta')]
print("FASTA files:", fasta_files)

# Part 4: Accepting a DNA sequence as a command-line argument
if len(sys.argv) > 1:
    input_sequence = sys.argv[1]
    print("Input sequence:", input_sequence)
else:
    print("No input sequence provided.")
````

#Efficiency and Optimization


**1. Big O Notation (Introductory)**

**Overview:**

Big O notation is a way to describe the efficiency of an algorithm in terms of time and space complexity. It helps you understand how an algorithm's performance scales as the size of the input data increases.

**a) Understanding Big O Notation**

- O(1): Constant time - the operation takes the same amount of time regardless of the input size.
- O(n): Linear time - the operation's time grows linearly with the input size.
- O(n^2): Quadratic time - the operation's time grows quadratically with the input size.

**Example:**

````python
# O(1) - Constant time example
def get_first_element(data):
    return data[0]

# O(n) - Linear time example
def sum_elements(data):
    total = 0
    for element in data:
        total += element
    return total

# O(n^2) - Quadratic time example
def print_all_pairs(data):
    for i in data:
        for j in data:
            print(i, j)
````

**Explanation:**

- O(1): The function get_first_element always takes the same time, no matter the size of data.
- O(n): The function sum_elements takes time proportional to the length of data.
- O(n^2): The function print_all_pairs takes time proportional to the square of the length of data.

**2. Optimizing Loops and Data Structures**

**a) Optimizing Loops**

Loops can often be optimized to improve performance, especially when working with large datasets.

````python
# Inefficient loop (O(n^2))
def inefficient_contains_duplicates(data):
    for i in range(len(data)):
        for j in range(i + 1, len(data)):
            if data[i] == data[j]:
                return True
    return False

# Optimized loop using a set (O(n))
def optimized_contains_duplicates(data):
    seen = set()
    for element in data:
        if element in seen:
            return True
        seen.add(element)
    return False
````

**Explanation:**

- Optimization: The optimized version uses a set for O(1) lookups, reducing the overall complexity from O(n^2) to O(n).

**b) Choosing the Right Data Structure**

Choosing the right data structure can significantly improve performance.

````python
# Example: Checking membership in a list vs. a set
data_list = [1, 2, 3, 4, 5]
data_set = {1, 2, 3, 4, 5}

# O(n) - Checking membership in a list
print(3 in data_list)

# O(1) - Checking membership in a set
print(3 in data_set)
````

**Explanation:**

- Lists vs. Sets: Checking membership in a list is O(n), whereas checking membership in a set is O(1), making sets a better choice for this operation.

**3. Memory Management**

**a) Python’s Memory Model**

Understanding how Python manages memory can help you write more efficient code.

- Reference Counting: Python uses reference counting to keep track of the number of references to an object in memory. When an object's reference count drops to zero, Python automatically deletes it.
- Garbage Collection: Python has a garbage collector that handles the cleanup of unused objects to free up memory.

**b) Tips for Managing Memory Effectively**

- Avoid Unnecessary Copies: Minimize the creation of unnecessary copies of large data structures.
- Use Generators: Generators allow you to iterate over data without storing the entire dataset in memory.

**Example:**

````python
# Inefficient: Creates a large list in memory
squares = [x**2 for x in range(1000000)]

# Efficient: Uses a generator, no large list in memory
squares_gen = (x**2 for x in range(1000000))
````

**Explanation:**

- Generators: The generator expression (x**2 for x in range(1000000)) does not create a large list in memory, making it more memory-efficient.

**Important:** Each participant appears once. Measurements before and after treatment are *paired within participants*; comparisons between different treatment groups are *independent*.

## 2. Inspect and prepare the data
Missing follow-up measurements are **not zero**. For tests requiring follow-up data, use only rows with nonmissing relevant measurements. Do not silently impute zero. Define change as follow-up minus baseline: a negative change means a decrease.

````python

# Check missing values and group sizes
print('Missing values by column:')
print(df.isna().sum())  # Count missing values (NaN) in each column

print('Group sizes:')
print(df['Group'].value_counts().sort_index())  # Count participants in each group, sorted alphabetically

# Convert selected columns to numeric data types
for col in ['Age_years', 'BP_Baseline_mmHg', 'BP_Followup_mmHg', 'Biomarker_mg_L']:
    df[col] = pd.to_numeric(df[col], errors='coerce')  # Convert invalid values to NaN

# Calculate blood pressure change (follow-up minus baseline)
df['BP_Change_mmHg'] = df['BP_Followup_mmHg'] - df['BP_Baseline_mmHg']

# Remove rows with missing baseline or follow-up BP values
complete = df.dropna(subset=['BP_Baseline_mmHg', 'BP_Followup_mmHg']).copy()
# .copy() creates an independent DataFrame without modifying the original df

# Calculate descriptive statistics of BP change for each group
print(complete.groupby('Group')['BP_Change_mmHg'].agg(['count', 'mean', 'std']).round(2))
# groupby(): Organize observations by treatment group
# agg(): Calculate sample count, mean, and standard deviation
# round(2): Display results rounded to two decimal places
````

## 4. Paired t-test: before vs. after within one group
**Question:** Did systolic blood pressure change within Treatment_A?

- H0: Mean within-person change = 0.
- H1: Mean within-person change ≠ 0.
- `ttest_rel` accounts for the paired nature of baseline/follow-up measurements.
- Check the distribution of within-person differences; the assumption is about differences, not baseline BP.
- A statistically significant change *within* one group does not by itself demonstrate the intervention is better than Control.

````python

# Select participants in Treatment_A
a = complete.loc[complete['Group'] == 'Treatment_A']

# Perform a paired t-test comparing follow-up and baseline BP in the same participants
t_stat, p_value = stats.ttest_rel(a['BP_Followup_mmHg'], a['BP_Baseline_mmHg'])
# ttest_rel(): Tests whether the mean difference between paired measurements is zero
# t_stat: t-test statistic; p_value: statistical significance of the test

# Calculate individual BP changes and their mean
diff = a['BP_Change_mmHg'].to_numpy()  # Convert BP changes to a NumPy array
mean_diff = diff.mean()                # Calculate the average BP change
se = stats.sem(diff)                   # Calculate the standard error of the mean change

# Calculate the 95% confidence interval (CI) for the mean BP change
ci = stats.t.interval(0.95, df=len(diff)-1, loc=mean_diff, scale=se)
# df: Degrees of freedom (number of paired observations minus 1)
# loc: Mean difference; scale: Standard error
# 95% CI: Range of plausible values for the population mean change

# Display statistical results
print(f'Paired t={t_stat:.3f}; p={p_value:.4g}')
print(f'Mean change={mean_diff:.2f} mmHg; 95% CI=({ci[0]:.2f}, {ci[1]:.2f})')

# Check the normality assumption of the paired differences
print('Shapiro-Wilk p for paired differences:', round(stats.shapiro(diff).pvalue, 4))
# shapiro(): Tests whether the paired differences are consistent with a normal distribution
# p < 0.05: Evidence against normality; p >= 0.05: No strong evidence against normality
````

## 5. Independent samples t-test: Treatment_A vs. Control
**Question:** Is the mean BP change different between independent groups?

- H0: Mean BP change for Treatment_A equals mean BP change for Control.
- Use **Welch's t-test**, which does not assume equal variances.
- The groups must consist of independent participants.

````python

# Select BP change values for Control and Treatment_A groups
control = complete.loc[complete['Group'] == 'Control', 'BP_Change_mmHg']
treat_a = complete.loc[complete['Group'] == 'Treatment_A', 'BP_Change_mmHg']
# .loc[]: Select BP change measurements for participants in each group

# Perform Welch's independent t-test to compare mean BP changes between groups
result = stats.ttest_ind(treat_a, control, equal_var=False)
# ttest_ind(): Compares the means of two independent groups
# equal_var=False: Uses Welch's t-test, which does not assume equal population variances

# Display the t-statistic and p-value
print(f'Welch t={result.statistic:.3f}; p={result.pvalue:.4g}')
# statistic: Calculated t-value; pvalue: Evidence against equal population means

# Calculate the difference between the two group means
print(f'Mean difference (A minus Control)={treat_a.mean()-control.mean():.2f} mmHg')
# Negative difference: Treatment_A has a greater average BP reduction than Control

# Calculate the 95% confidence interval for the difference in group means
print('95% CI:', result.confidence_interval(confidence_level=0.95))
# confidence_interval(): Estimates a 95% CI for the difference (Treatment_A minus Control)
# A 95% CI excluding zero corresponds to statistical significance at alpha = 0.05
# for this two-sided Welch t-test
````

## 6. One-way ANOVA: three independent groups
**Question:** Is the average BP change the same across Control, Treatment_A and Treatment_B?

- H0: All three group means are equal.
- H1: At least one group mean differs.
- Classic ANOVA assumes independent observations, approximately normal residuals, and similar variances.
- Check a variance diagnostic. If equal variance is implausible, use Welch ANOVA instead.
- A significant omnibus test does **not** tell us which specific groups differ.

````python

# Extract BP change values for each study group as NumPy arrays
control_vals, a_vals, b_vals = [
    complete.loc[complete['Group'] == g, 'BP_Change_mmHg'].to_numpy()
    for g in groups
]
# .loc[]: Select BP change values for each group
# .to_numpy(): Convert pandas data into NumPy arrays

# Check whether the groups have similar variances
print('Levene (median-centered) test:', stats.levene(control_vals, a_vals, b_vals, center='median'))
# Levene's test: Tests whether group variances differ
# center='median': Uses the median (Brown-Forsythe variation), which is more robust to outliers
# p < 0.05 suggests unequal variances

# Perform a one-way ANOVA to compare mean BP changes across three groups
print('Classic one-way ANOVA:', stats.f_oneway(control_vals, a_vals, b_vals))
# f_oneway(): Calculates the F-statistic and p-value
# p < 0.05 suggests that at least one group mean differs

# Fit a linear model for ANOVA using statsmodels
model = ols('BP_Change_mmHg ~ C(Group)', data=complete).fit()
# ols(): Fits an ordinary least squares model
# ~ : Separates the outcome (BP change) from the predictor (Group)
# C(Group): Treats Group as a categorical variable
# .fit(): Estimates the model parameters

# Display the ANOVA table
print(sm.stats.anova_lm(model, typ=2))
# anova_lm(): Produces an ANOVA table with F-statistic, degrees of freedom, and p-value
# typ=2: Uses Type II sums of squares

# Perform Tukey's HSD test to identify which groups differ
print(pairwise_tukeyhsd(endog=complete['BP_Change_mmHg'], groups=complete['Group'], alpha=0.05))
# pairwise_tukeyhsd(): Compares every pair of group means while adjusting for multiple comparisons
# endog: Numerical outcome variable (BP change)
# groups: Categorical variable defining the groups
# alpha=0.05: Significance level of 5%
# Note: Standard Tukey HSD assumes independent observations and equal population variances
````

## 7. Categorical data: chi-square test
**Question:** Is the fraction of participants meeting the responder definition related to treatment group?

Here a responder is defined in this *fictional exercise* as BP change ≤ -10 mmHg. The indicator was generated before the missing follow-up data were inserted. **Recalculate responder status using the observed change, so no response is used when follow-up BP is missing.**

- H0: Response status is independent of study group.
- Check expected cell counts; for sparse tables consider an exact/alternative test.

````python

# Create a responder variable based on blood pressure reduction
complete['Responder_Recomputed'] = np.where(complete['BP_Change_mmHg'] <= -10, 'Yes', 'No')
# np.where(): Assigns 'Yes' if BP decreased by at least 10 mmHg; otherwise assigns 'No'

# Create a contingency table of responders by treatment group
table = pd.crosstab(complete['Group'], complete['Responder_Recomputed'])
# pd.crosstab(): Counts the number of 'Yes' and 'No' responses in each group
print(table)

# Perform a chi-square test of independence
chi2, p, dof, expected = stats.chi2_contingency(table)
# chi2: Chi-square test statistic
# p: P-value testing the association between group and responder status
# dof: Degrees of freedom
# expected: Expected counts under the assumption of independence

# Display the statistical results
print(f'Chi-square={chi2:.3f}, df={dof}, p={p:.4g}')

# Check the smallest expected cell count
print('Smallest expected cell count:', expected.min())
# Expected counts should generally be at least 5 for the usual chi-square approximation
# p < 0.05 suggests a statistically significant association between group and response
````


## 8. Pearson and Spearman Correlation

**Question:** Are age and baseline systolic BP associated?

- Calculate Pearson and Spearman correlation coefficients.
- Create a scatter plot of age versus baseline systolic BP.
- Add a linear trend line and compare it with the correlation results.
- Interpret the direction, strength, and statistical significance of the association. Remember that correlation does not imply causation.


````python

# Select age and baseline systolic BP, excluding rows with missing values
corr_data = df[['Age_years', 'BP_Baseline_mmHg']].dropna()
# dropna(): Remove rows with missing values in either variable

# Calculate Pearson correlation
pearson_r, pearson_p = stats.pearsonr(corr_data['Age_years'], corr_data['BP_Baseline_mmHg'])
# pearsonr(): Measures the strength and direction of a linear relationship
# r ranges from -1 (perfect negative) to +1 (perfect positive)

# Calculate Spearman correlation
spearman_r, spearman_p = stats.spearmanr(corr_data['Age_years'], corr_data['BP_Baseline_mmHg'])
# spearmanr(): Measures a rank-based monotonic relationship
# Spearman is less sensitive to extreme values than Pearson

# Display correlation coefficients and p-values
print(f'Pearson correlation: r={pearson_r:.3f}, p={pearson_p:.4g}')
print(f'Spearman correlation: rho={spearman_r:.3f}, p={spearman_p:.4g}')
# p < 0.05 suggests a statistically significant association
# Correlation does not establish causation

# Create a scatter plot of age versus baseline systolic BP
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(corr_data['Age_years'], corr_data['BP_Baseline_mmHg'], alpha=0.7)
# scatter(): Display each participant as a point
# alpha=0.7: Make overlapping points easier to see

# Add a linear trend line
slope, intercept = np.polyfit(corr_data['Age_years'], corr_data['BP_Baseline_mmHg'], 1)
# polyfit(): Fit a straight line (degree 1) to the data

x = np.linspace(corr_data['Age_years'].min(), corr_data['Age_years'].max(), 100)
# linspace(): Generate 100 evenly spaced age values

ax.plot(x, slope * x + intercept, linestyle='--', label='Linear trend')
# plot(): Draw the fitted trend line

# Add labels and title
ax.set_xlabel('Age (years)')
ax.set_ylabel('Baseline systolic BP (mmHg)')
ax.set_title('Age vs. Baseline Systolic Blood Pressure')
ax.legend()

plt.tight_layout()  # Adjust spacing
plt.show()          # Display the figure
````

## 9. Optional: linear regression with covariates
For randomized group comparisons, a baseline-adjusted follow-up model can account for initial BP and age. The coefficients are associations conditional on the included covariates. Proper causal interpretation depends on the study design; this synthetic dataset is not a clinical trial.

````python

# Prepare the dataset for multiple linear regression
reg = df.dropna(subset=['BP_Followup_mmHg', 'BP_Baseline_mmHg', 'Age_years']).copy()
# dropna(): Remove rows with missing values in the specified columns
# .copy(): Create an independent DataFrame for regression analysis

# Fit a multiple linear regression model
fit = ols('BP_Followup_mmHg ~ C(Group) + BP_Baseline_mmHg + Age_years', data=reg).fit()
# ols(): Ordinary least squares regression
# BP_Followup_mmHg: Dependent variable (outcome to predict)
# C(Group): Categorical predictor representing treatment groups
# BP_Baseline_mmHg: Numerical predictor representing baseline blood pressure
# Age_years: Numerical predictor representing participant age
# + : Includes multiple predictors in the regression model
# .fit(): Estimates the regression coefficients

# Display the regression results
print(fit.summary())
# summary(): Shows regression coefficients, standard errors, t-statistics, p-values,
#            95% confidence intervals, R-squared, and overall model statistics
# Each group coefficient compares that group with the reference group,
# adjusting for baseline BP and age
````

````python

# ============================================================
# MDE-8084 — ASSIGNMENT 4: COMPLETE INSTRUCTOR SOLUTION
# Independent Statistical Analysis in Python
# Fictional observational rehabilitation study
# All data are SYNTHETIC.
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy import stats
from itertools import combinations
from statsmodels.stats.multitest import multipletests
from statsmodels.stats.oneway import anova_oneway
import statsmodels.formula.api as smf

ALPHA = 0.05
FILE = "assignment4_new_rehabilitation_data.csv"

programs = [
    "Standard_Care",
    "Supervised_Rehab",
    "Digital_Rehab"
]

def heading(title):
    print("\n" + "=" * 75)
    print(title)
    print("=" * 75)

def decision(p):
    if p < ALPHA:
        return "Reject H0 (statistically significant)."
    return "Do not reject H0 (not statistically significant)."

# Load the file, using Colab upload if necessary.
if not os.path.exists(FILE):
    from google.colab import files
    print("Please upload:", FILE)
    files.upload()

df = pd.read_csv(FILE)


# ============================================================
# PART A — UNDERSTAND THE DATA
# ============================================================

# Question 1: Data inspection

heading("PART A — Q1: Data inspection")

print("Dataset shape:", df.shape)
print("\nData types:\n", df.dtypes)
print("\nFirst five rows:\n", df.head())
print("\nMissing values:\n", df.isna().sum())

duplicates = df["Participant_ID"].duplicated().sum()
print("\nDuplicate IDs:", duplicates)

print("\nParticipants in each program:")
print(df["Program"].value_counts().reindex(programs))

if duplicates:
    raise ValueError("Duplicate IDs require investigation.")


# Question 2: Calculate walking-distance change

heading("PART A — Q2: Walking-distance change")

df["Walk_Change_m"] = (
    df["Walk_Week8_m"] - df["Walk_Baseline_m"]
)

# Participants with both measurements.
complete = df.dropna(
    subset=["Walk_Baseline_m", "Walk_Week8_m"]
).copy()

print("Total participants:", len(df))
print("Complete walking-distance pairs:", len(complete))
print("Excluded from change-based analyses:",
      len(df) - len(complete))

print("\nComplete pairs by group:")
print(complete["Program"].value_counts().reindex(programs))

print("\nMissing follow-up by group:")
print(
    df.groupby("Program")["Walk_Week8_m"]
      .apply(lambda x: x.isna().sum())
      .reindex(programs)
)

print("\nMissing-data strategy:")
print("Do not replace missing follow-up values with zero.")
print("Use complete pairs for analyses of walking change.")
print("Use available baseline data for baseline-only analyses.")
print("Missing follow-up may introduce selection bias.")


# Question 3: Descriptive statistics

heading("PART A — Q3: Descriptive statistics")

variables = [
    "Age_years",
    "BMI_kg_m2",
    "Walk_Baseline_m",
    "Walk_Week8_m",
    "Walk_Change_m"
]

summary = (
    df.groupby("Program")[variables]
      .agg(["count", "mean", "std", "median"])
      .reindex(programs)
)

print(summary.round(2).to_string())

group_means = (
    complete.groupby("Program")["Walk_Change_m"]
            .mean()
            .reindex(programs)
)

print("\nMean change by group (m):")
print(group_means.round(2))

print("\nLargest descriptive improvement:",
      group_means.idxmax())

print("These are associations, not causal effects.")


# ============================================================
# PART B — EXPLORE VISUALLY
# ============================================================

# Question 4: Box plots of walking-distance change

heading("PART B — Q4: Box plot")

group_data = [
    complete.loc[
        complete["Program"] == group,
        "Walk_Change_m"
    ].to_numpy()
    for group in programs
]

fig, ax = plt.subplots(figsize=(9, 5))

ax.boxplot(
    group_data,
    tick_labels=programs,
    showmeans=True
)

ax.axhline(
    y=0,
    linestyle="--",
    color="gray",
    label="No change"
)

ax.set_xlabel("Rehabilitation program")
ax.set_ylabel("Walking-distance change (m)")
ax.set_title("Eight-Week Change in Walking Distance")
ax.legend()

fig.tight_layout()
plt.show()

print("Interpretation:")
print("Positive change indicates improved walking distance.")
print("Compare medians, variability, and outliers.")
print("The programs may differ in their typical improvement.")


# Question 5: Age versus baseline walking distance

heading("PART B — Q5: Scatter plot")

baseline = df[
    ["Age_years", "Walk_Baseline_m"]
].dropna()

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(
    baseline["Age_years"],
    baseline["Walk_Baseline_m"],
    alpha=0.65
)

ax.set_xlabel("Age (years)")
ax.set_ylabel("Baseline walking distance (m)")
ax.set_title("Age vs. Baseline Walking Distance")

fig.tight_layout()
plt.show()

print("Examine whether the pattern is approximately linear,")
print("monotonic, or unclear; look for possible outliers.")


# ============================================================
# PART C — CHOOSE AND APPLY STATISTICAL TESTS
# ============================================================

# Question 6: Paired t-test, Supervised_Rehab

heading("PART C — Q6: Paired t-test")

supervised = complete[
    complete["Program"] == "Supervised_Rehab"
]

changes = supervised["Walk_Change_m"].to_numpy()

paired_test = stats.ttest_rel(
    supervised["Walk_Week8_m"],
    supervised["Walk_Baseline_m"]
)

n = len(changes)
mean_change = changes.mean()
sd_change = changes.std(ddof=1)

ci_paired = stats.t.interval(
    confidence=0.95,
    df=n - 1,
    loc=mean_change,
    scale=stats.sem(changes)
)

cohen_dz = mean_change / sd_change

print("H0: Mean within-person change = 0 m.")
print("H1: Mean within-person change != 0 m.")

print("Why paired?")
print("Each participant has baseline and week-8 measurements.")

print(f"\nN = {n}")
print(f"Mean change = {mean_change:.3f} m")
print(f"t({n-1}) = {paired_test.statistic:.3f}")
print(f"p = {paired_test.pvalue:.6g}")

print(
    f"95% CI = "
    f"[{ci_paired[0]:.3f}, {ci_paired[1]:.3f}] m"
)

print(f"Cohen's dz = {cohen_dz:.3f}")
print(decision(paired_test.pvalue))


# Question 7: Welch independent t-test

heading("PART C — Q7: Independent Welch t-test")

digital = group_data[2]
standard = group_data[0]

welch = stats.ttest_ind(
    digital,
    standard,
    equal_var=False
)

mean_difference = digital.mean() - standard.mean()

a = digital.var(ddof=1) / len(digital)
b = standard.var(ddof=1) / len(standard)

se_difference = np.sqrt(a + b)

welch_df = (a + b)**2 / (
    a**2 / (len(digital) - 1)
    + b**2 / (len(standard) - 1)
)

ci_welch = stats.t.interval(
    confidence=0.95,
    df=welch_df,
    loc=mean_difference,
    scale=se_difference
)

print("H0: Both programs have equal mean walking changes.")
print("H1: Their mean walking changes differ.")

print("\nWhy Welch independent t-test?")
print("Digital_Rehab and Standard_Care have different")
print("participants, so their measurements are independent.")
print("Welch's method does not assume equal variances.")

print(f"\nMean difference = {mean_difference:.3f} m")
print(f"t({welch_df:.2f}) = {welch.statistic:.3f}")
print(f"p = {welch.pvalue:.6g}")

print(
    f"95% CI = "
    f"[{ci_welch[0]:.3f}, {ci_welch[1]:.3f}] m"
)

print(decision(welch.pvalue))


# Question 8: One-way ANOVA

heading("PART C — Q8: One-way ANOVA")

anova = stats.f_oneway(*group_data)

N = sum(len(x) for x in group_data)
k = len(group_data)

df_between = k - 1
df_within = N - k

print("H0: All three population mean changes are equal.")
print("H1: At least one mean differs.")

print(f"\nF({df_between}, {df_within}) = "
      f"{anova.statistic:.3f}")

print(f"p = {anova.pvalue:.6g}")
print(decision(anova.pvalue))


# Check variance assumption: Brown-Forsythe

levene = stats.levene(
    *group_data,
    center="median"
)

print("\nBrown-Forsythe test for variance differences:")
print(f"Statistic = {levene.statistic:.3f}")
print(f"p = {levene.pvalue:.6g}")

if levene.pvalue < ALPHA:
    print("Evidence that group variances differ.")
else:
    print("No strong evidence that group variances differ.")
    print("This does not prove equal variances.")


# Welch ANOVA as a robustness check

welch_anova = anova_oneway(
    group_data,
    use_var="unequal"
)

print("\nWelch ANOVA:")
print(f"F = {welch_anova.statistic:.3f}")
print(f"df = {welch_anova.df}")
print(f"p = {welch_anova.pvalue:.6g}")


# ANOVA effect size: eta-squared

all_changes = np.concatenate(group_data)
grand_mean = all_changes.mean()

ss_between = sum(
    len(x) * (x.mean() - grand_mean)**2
    for x in group_data
)

ss_total = np.sum((all_changes - grand_mean)**2)

eta_squared = ss_between / ss_total

print(f"\nEta-squared = {eta_squared:.3f}")


# Pairwise comparisons: Welch tests + Holm correction

comparisons = []

for group1, group2 in combinations(programs, 2):

    x = complete.loc[
        complete["Program"] == group1,
        "Walk_Change_m"
    ].to_numpy()

    y = complete.loc[
        complete["Program"] == group2,
        "Walk_Change_m"
    ].to_numpy()

    test = stats.ttest_ind(
        x, y, equal_var=False
    )

    comparisons.append({
        "Comparison": f"{group1} vs {group2}",
        "Mean_Difference_m": x.mean() - y.mean(),
        "Raw_P": test.pvalue
    })

pair_df = pd.DataFrame(comparisons)

pair_df["Holm_P"] = multipletests(
    pair_df["Raw_P"],
    method="holm"
)[1]

pair_df["Significant"] = pair_df["Holm_P"] < ALPHA

print("\nPost-hoc results:")
print(pair_df.round(5).to_string(index=False))

print("\nOther assumptions:")
print("- Independent participants")
print("- Approximately normal residuals within groups")
print("- Reasonable handling of outliers")
print("- Equal variances for classical ANOVA")
print("Recruitment independence cannot be proven by a p-value.")


# Question 9: Chi-square test of responders

heading("PART C — Q9: Chi-square test")

# A gain of >=30 meters is an illustrative threshold,
# not a universal clinical cutoff.

complete["Responder"] = np.where(
    complete["Walk_Change_m"] >= 30,
    "Yes",
    "No"
)

contingency = pd.crosstab(
    complete["Program"],
    complete["Responder"]
).reindex(
    index=programs,
    columns=["No", "Yes"],
    fill_value=0
)

chi2, chi_p, chi_df, expected = (
    stats.chi2_contingency(
        contingency,
        correction=False
    )
)

print("Observed counts:")
print(contingency)

print("\nExpected counts:")
print(
    pd.DataFrame(
        expected,
        index=contingency.index,
        columns=contingency.columns
    ).round(2)
)

print(f"\nChi-square({chi_df}) = {chi2:.3f}")
print(f"p = {chi_p:.6g}")
print(f"Minimum expected count = {expected.min():.2f}")

print(decision(chi_p))

if expected.min() < 5:
    print("Warning: Small expected counts may make")
    print("the chi-square approximation unreliable.")
else:
    print("Expected cell counts support the usual")
    print("chi-square approximation.")

print("Independence of participants is also required.")


# ============================================================
# PART D — INTERPRET, DON'T JUST CALCULATE
# ============================================================

# Question 10: Pearson and Spearman correlations

heading("PART D — Q10: Correlations")

pearson = stats.pearsonr(
    baseline["Age_years"],
    baseline["Walk_Baseline_m"]
)

spearman = stats.spearmanr(
    baseline["Age_years"],
    baseline["Walk_Baseline_m"]
)

print("Pearson correlation:")
print(f"r = {pearson.statistic:.3f}")
print(f"p = {pearson.pvalue:.6g}")

print("\nSpearman correlation:")
print(f"rho = {spearman.statistic:.3f}")
print(f"p = {spearman.pvalue:.6g}")

print("\nPearson measures linear association.")
print("Spearman measures monotonic rank association.")
print("Neither establishes causation.")


# Question 11: Non-significant results

heading("PART D — Q11: Non-significant results")

primary_pvalues = {
    "Paired t-test": paired_test.pvalue,
    "Independent Welch t-test": welch.pvalue,
    "ANOVA": anova.pvalue,
    "Chi-square": chi_p,
    "Pearson correlation": pearson.pvalue,
    "Spearman correlation": spearman.pvalue
}

non_significant = [
    (name, p)
    for name, p in primary_pvalues.items()
    if p >= ALPHA
]

if non_significant:
    name, p = non_significant[0]

    print("Example:", name)
    print(f"p = {p:.5g}")
    print("Insufficient evidence to reject H0.")
    print("This is not proof of zero effect.")
else:
    print("All primary tests are statistically significant.")
    print("A hypothetical p >= 0.05 would indicate")
    print("insufficient evidence against H0.")
    print("It would not establish equivalence or no effect.")

print("\nEquivalence requires a prespecified equivalence")
print("margin and an appropriate equivalence test.")


# Question 12: Scientific interpretation

heading("PART D — Q12: Scientific interpretation")

interpretation_text = f"""
In this fictional observational study, average eight-week
walking-distance change varied across rehabilitation programs.
The largest descriptive improvement was seen in
{group_means.idxmax()} ({group_means.max():.1f} m), but group
differences must not be interpreted as treatment effects.
Statistical significance measures how compatible the observed
data are with a null hypothesis under a specified model; it
does not establish whether the estimated distance change
matters to patients. Clinical importance would require a
justified, population-specific threshold and attention to
confidence intervals, variability, and patient goals.
Because participants were not randomized, differences in
severity, motivation, access, or other unmeasured
characteristics could confound program comparisons, even
after regression adjustment. The data are entirely simulated
and provide no evidence that any real rehabilitation program
is effective. Twelve participants had missing week-eight
measurements. Excluding them assumes that the available cases
adequately represent the intended comparison, which may be
incorrect if follow-up depends on health status or response.
The sample also may not represent broader clinical
populations. These analyses should therefore be considered
programming and statistical practice, not a basis for clinical
recommendations.
"""

print(interpretation_text)
print("Word count:", len(interpretation_text.split()))


# ============================================================
# PART E — EXTEND YOUR ANALYSIS
# ============================================================

# Question 13: Multiple linear regression

heading("PART E — Q13: Multiple regression")

regression_df = df.dropna(
    subset=[
        "Walk_Week8_m",
        "Walk_Baseline_m",
        "Age_years",
        "BMI_kg_m2",
        "Program"
    ]
).copy()

formula = (
    'Walk_Week8_m ~ '
    'C(Program, Treatment(reference="Standard_Care")) '
    '+ Walk_Baseline_m + Age_years + BMI_kg_m2'
)

# HC3: heteroskedasticity-robust standard errors.
model = smf.ols(
    formula,
    data=regression_df
).fit(cov_type="HC3")

print("Number of observations:", len(regression_df))
print(model.summary())

term = (
    'C(Program, Treatment(reference="Standard_Care"))'
    '[T.Supervised_Rehab]'
)

coefficient = model.params[term]
ci = model.conf_int().loc[term]

print("\nInterpretation of Supervised_Rehab coefficient:")
print(f"Adjusted coefficient = {coefficient:.3f} m")
print(f"95% CI = [{ci.iloc[0]:.3f}, {ci.iloc[1]:.3f}] m")
print(f"p = {model.pvalues[term]:.6g}")

print("\nHolding age, BMI, and baseline walking distance")
print("constant, the model estimates a difference of")
print(f"{coefficient:.2f} meters in week-8 walking distance")
print("between Supervised_Rehab and Standard_Care.")
print("This association is not necessarily causal.")


# Regression diagnostic plot

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(
    model.fittedvalues,
    model.resid,
    alpha=0.65
)

ax.axhline(
    0,
    linestyle="--",
    color="gray"
)

ax.set_xlabel("Predicted week-8 walking distance (m)")
ax.set_ylabel("Regression residual (m)")
ax.set_title("Regression Residuals vs Fitted Values")

fig.tight_layout()
plt.show()

print("\nRegression diagnostic considerations:")
print("- Linear/additive relationships")
print("- Constant error variance")
print("- Residual distribution and influential observations")
print("- Multicollinearity")
print("- Independence of observations")
print("- Possible unmeasured confounding")

print("HC3 robust errors address heteroskedasticity")
print("in standard errors, not omitted-variable bias.")


# Question 14: Multiple-testing correction

heading("PART E — Q14: Multiple testing")

print("Pairwise comparisons after Holm correction:")

pair_df["Unadjusted_Significant"] = (
    pair_df["Raw_P"] < ALPHA
)

pair_df["Holm_Significant"] = (
    pair_df["Holm_P"] < ALPHA
)

print(
    pair_df[
        [
            "Comparison",
            "Raw_P",
            "Holm_P",
            "Unadjusted_Significant",
            "Holm_Significant"
        ]
    ].round(6).to_string(index=False)
)

changed = pair_df[
    pair_df["Unadjusted_Significant"]
    != pair_df["Holm_Significant"]
]

if changed.empty:
    print("\nNo pairwise significance conclusions changed.")
else:
    print("\nConclusions changed for:")
    print(changed["Comparison"].to_string(index=False))

print("\nWhy adjust?")
print("Testing multiple hypotheses increases the chance")
print("of false-positive findings.")

print("\nHolm:")
print("Controls the family-wise error rate.")

print("\nBenjamini-Hochberg:")
print("Controls the false discovery rate and may be useful")
print("for exploratory analyses involving many comparisons,")
print("such as testing thousands of genes.")


# ============================================================
# FINISHED
# ============================================================

heading("ASSIGNMENT 4 COMPLETE")

print("All 14 tasks have been addressed.")
print("Data source: synthetic rehabilitation dataset.")
print("Do not interpret findings as real clinical evidence.")
````
