# Introduction to Functions

A **function** is a named, reusable block of code that performs a specific task. Instead of repeating instructions, we define them once and call the function with different inputs.

Functions help us:

- **Reuse code:** Apply the same operation to different sequences or patient records.
- **Organize code:** Divide a larger analysis into smaller tasks.
- **Test and maintain code:** Check individual tasks and make corrections in one place.

## a) Defining Functions

A function can receive inputs, process them, and return a result for use elsewhere in the program.

### Syntax

```python
def function_name(parameter):
    # Instructions for the task
    return result
```

### Understanding the Syntax

- **`def`:** Begins a function definition.
- **`function_name`:** A descriptive name, usually written in lowercase with underscores.
- **`parameter`:** A variable that receives an input when the function is called. A function can have zero, one, or multiple parameters.
- **`:`:** Marks the beginning of the function body.
- **Indentation:** Groups the instructions inside the function, typically using four spaces.
- **`return`:** Sends a result back to the caller. Not every function needs an explicit `return` statement.

### Important Notes

- **Defining a function does not run its body.** We must call the function to execute its instructions.
- **Parameters** appear in the definition; **arguments** are the actual values supplied when calling it.
- **`return` and `print()` serve different purposes:** `return` provides a result that other code can use; `print()` displays information.

### Example: Turn Our Week 1 GC Calculation into a Function

In Week 1, we calculated GC content directly in a script. Now we will put that familiar calculation inside `calculate_gc_content()` so we can reuse it with different DNA sequences.

Focus on the new structure: the function receives a sequence through a parameter and sends the calculated percentage back using `return`.

For this first example, assume the input is a nonempty, uppercase sequence containing only A, T, G, and C. We will add input validation later.

````python
# Turn our Week 1 calculation into a reusable function.
# Assumes a nonempty, uppercase DNA sequence containing only A, T, G, and C.
def calculate_gc_content(dna_sequence):
    # dna_sequence receives the input supplied when we call the function.
    g_count = dna_sequence.count('G')
    c_count = dna_sequence.count('C')
    gc_content = (g_count + c_count) / len(dna_sequence) * 100

    # Send the result back so other code can store or use it.
    return gc_content
````

### Explanation

- **Function name — `calculate_gc_content`:** We chose this name to describe the task our function performs. Using `def` creates our own function, which we can reuse with different DNA sequences.

- **Parameter — `dna_sequence`:** This variable receives the sequence supplied when we call the function. Inside the function, we use this parameter to work with the input.

- **String method — `.count()`:** A method is a function associated with an object. Here, `dna_sequence` is a string, and `dna_sequence.count('G')` calls its `.count()` method to count uppercase `G` characters. Similarly, `.count('C')` counts uppercase `C` characters. The dot connects the string to its method.

- **Built-in function — `len()`:** Python provides `len()` to determine the length of an object. For this string, `len(dna_sequence)` returns the number of characters. Notice the different syntax: we write `len(dna_sequence)`, whereas the method is written `dna_sequence.count('G')`.

- **Calculation:** We add the G and C counts, divide by the sequence length, and multiply by 100. The percentage is stored in `gc_content`.

- **Return statement — `return gc_content`:** `return` is a Python keyword that ends the function call and sends the result back to the caller. Other code can store, print, or use that result. The statement itself does not print anything.

**Important:** We have only defined the function. Its calculations run when we call it, which we will do next.

## b) Calling Functions

Once we have defined a function, we can **call (execute)** it to perform the task it was designed to do.

Calling a function means using its name followed by parentheses `()`. If the function requires input values, we provide them inside the parentheses. These values are called **arguments**.

When a function is called, Python executes the instructions inside the function using the provided arguments. If the function contains a `return` statement, the returned value can be stored in a variable or used in further calculations.

**Remember:**
- **Defining a function:** Tells Python what the function should do.
- **Calling a function:** Tells Python to execute the function using the provided arguments.

### Syntax:

```python
function_name(arguments)
```

### Understanding the Syntax:

- **`function_name`:** The name of the function we want to execute. It must refer to a function that has already been defined or is otherwise available in Python.

- **`arguments`:** The actual values or variables we provide when calling the function. These values are passed to the function's parameters so that it can perform its task.

- **Parentheses `()`:** Used to call the function. Even if a function does not require arguments, we must still include empty parentheses when calling it.

### Example: Calling the GC Content Function

In the previous section, we defined a function named `calculate_gc_content()` that calculates the percentage of guanine (G) and cytosine (C) nucleotides in a DNA sequence.

Now, we will **call this function** using a specific DNA sequence and display the calculated GC content.

In this example:

1. We create a variable named `sequence` to store a DNA sequence.
2. We pass `sequence` as an argument to `calculate_gc_content()`.
3. Python executes the function and calculates the GC percentage.
4. The function returns the calculated value, which we store in `gc_content`.
5. We use `print()` to display the result.

**Important:** The function must be defined before we call it. In Google Colab, make sure you have already run the previous code cell containing the function definition.

````python

# Example: Calling the GC Content Function

# Create a variable named 'sequence' to store a DNA sequence.
# The sequence contains the four DNA nucleotides: A, T, G, and C.
sequence = "ATGCGCGATCGTACGCTAGC"

# Call the previously defined calculate_gc_content() function.
# Pass 'sequence' as an argument to the function.
# Python uses this value as the 'dna_sequence' parameter inside the function.
# The function calculates the GC percentage and returns the result.
# Store the returned value in a new variable named 'gc_content'.
gc_content = calculate_gc_content(sequence)

# Display the calculated GC content using the print() function.
# The letter 'f' before the string creates an f-string (formatted string).
# {gc_content:.2f} inserts the value of gc_content into the string.
# The :.2f formatting displays the number with exactly two decimal places.
# The % symbol is displayed after the calculated value.
print(f"GC Content: {gc_content:.2f}%")
````

### Explanation:

- **Calling the Function:** The function `calculate_gc_content()` is called with the DNA sequence `"ATGCGCGATCGTACGCTAGC"` as its argument. The variable `sequence` holds this DNA sequence and passes it to the function, where it is received by the parameter `dna_sequence`.

- **Storing the Result:** The function calculates the percentage of G and C nucleotides and returns the result using the `return` statement. This returned value is stored in the variable `gc_content`, allowing us to use it later without repeating the calculation.

- **Printing the Result:** The `print()` function displays the calculated GC content. We use an **f-string** (`f"..."`) to insert the value of `gc_content` directly into the text. The formatting expression `:.2f` displays the result with exactly two decimal places, producing `60.00%`.

**Key Takeaway:** A function can receive data through arguments, process that data, and return a result. By storing the returned value in a variable, we can display it, reuse it in other calculations, or pass it to another function.

## c) Parameters and Return Values

### Parameters

**Parameters** are variables defined inside the parentheses `()` of a function. They act as placeholders for the input values that a function receives when it is called.

Parameters allow us to write **flexible and reusable functions** that can work with different inputs without changing the function's code.

For example, instead of creating a separate function for every person or DNA sequence, we can define one function that accepts different values through its parameters.

A function can have **no parameters, one parameter, or multiple parameters**, depending on the task it needs to perform.

**Important Concepts:**

- **Parameters:** Variables listed in the function definition that represent the expected inputs.
- **Arguments:** The actual values passed to the function when it is called.
- **Multiple Parameters:** A function can accept more than one input, with parameters separated by commas (`,`).
- **Parameter Order:** When passing positional arguments, Python assigns values to parameters in the order they are provided.

### Example with Multiple Parameters

In the following example, we will define a function named `greet()` that accepts two parameters:

- `name`: Represents a person's name.
- `age`: Represents the person's age.

The function will use these parameters to display a personalized greeting.

We will then call the same function with different arguments to demonstrate how **one function can produce different outputs depending on the values provided**.

````python

# Example: Defining and Calling a Function with Multiple Parameters

# Define a function named greet() that accepts two parameters:
# 'name' represents the person's name.
# 'age' represents the person's age.
def greet(name, age):

    # Use print() to display a personalized greeting.
    # The 'f' before the string indicates an f-string (formatted string).
    # {name} is replaced with the value passed to the 'name' parameter.
    # {age} is replaced with the value passed to the 'age' parameter.
    print(f"Hello, {name}! You are {age} years old.")


# Call the greet() function with different arguments.

# First function call:
# "Alice" is passed to the 'name' parameter.
# 30 is passed to the 'age' parameter.
# The function executes and prints a greeting for Alice.
greet("Alice", 30)

# Second function call:
# "Bob" is passed to the 'name' parameter.
# 25 is passed to the 'age' parameter.
# The same function executes again using the new arguments.
greet("Bob", 25)
````

### Return Values

The **`return` statement** is used to send a result back to the part of the program that called a function.

When a function performs a calculation or processes data, we often want to use the result later in our program. The `return` statement allows us to retrieve that result and store it in a variable or use it in another operation.

A function can return different types of data, including **numbers, strings, lists, tuples, and dictionaries**.

If a function does not contain a `return` statement, or finishes executing without reaching one, Python automatically returns `None`, which represents the absence of a value.

### Important Concepts

- **`return`:** Sends a value back to the caller of the function.
- **Returned Value:** The result produced by the function, which can be stored in a variable or used in further calculations.
- **`None`:** The default value returned when a function finishes without explicitly returning a result.
- **Function Execution:** When Python executes a `return` statement, the function immediately stops executing and sends the specified value back to the caller.

### Difference Between `return` and `print()`

Although both are commonly used inside functions, they serve different purposes:

- **`print()`:** Displays information on the screen but does not make that information available as the function's return value.
- **`return`:** Sends a value back to the caller so it can be stored, reused, or processed further. It does not automatically display the value on the screen.

For example, in medical data analysis, we may write a function that calculates the average of several laboratory measurements. By returning the calculated average, we can use it in additional calculations or analyses without repeating the original computation.

### Example: Returning the Sum of Two Numbers

In the following example, we will define a function named `add_numbers()` that accepts two parameters, `a` and `b`.

The function will:

1. Receive two numbers as arguments.
2. Add the numbers together.
3. Use `return` to send the calculated sum back to the caller.
4. Store the returned value in a variable named `result`.
5. Display the result using `print()`.

This example demonstrates how **parameters allow us to provide input to a function, while `return` allows us to retrieve its output**.

````python

# Example: Using a Function with a Return Value

# Define a function named add_numbers() that accepts two parameters.
# 'a' represents the first number.
# 'b' represents the second number.
def add_numbers(a, b):

    # Add the values of 'a' and 'b' together.
    # The return statement sends the calculated sum back to the caller.
    # Unlike print(), return does not display the result on the screen.
    # Once return is executed, the function stops running.
    return a + b


# Call the add_numbers() function with two arguments: 5 and 3.
# The value 5 is assigned to parameter 'a'.
# The value 3 is assigned to parameter 'b'.
# The function calculates 5 + 3 and returns 8.
# Store the returned value (8) in the variable named 'result'.
result = add_numbers(5, 3)

# Display the returned result using the print() function.
# The text "The sum is:" is displayed first.
# The comma separates the text from the value stored in 'result'.
# Python automatically adds a space between them.
print("The sum is:", result)
````

### Key Points:

- **Parameters:** Variables defined in a function that allow it to receive input data. They make functions flexible and reusable with different inputs.

- **Arguments:** The actual values passed to a function when it is called. These values are assigned to the corresponding parameters.

- **Return Values:** Results sent back to the caller using the `return` statement. Returned values can be stored in variables, used in calculations, or passed to other functions.

- **`return` vs. `print()`:** The `return` statement sends a result back to the caller, while `print()` displays information on the screen.

- **Function Reusability:** Once a function is defined, it can be called multiple times with different arguments without rewriting its code.

**Remember:** Parameters allow information to enter a function, while return values allow information to come back out. Together, they help us write organized, reusable, and efficient Python programs.

### Exercise: Finding the First Stop Codon in a DNA Sequence

**Objective:** Convert the stop-codon search from Week 1 into a reusable function that returns a result.

### Connection to Week 1

In Week 1, we used a `while` loop to search a DNA sequence and print the position of a stop codon. Now, we will place that search inside a function so we can use it with different sequences.

The new programming concepts are:

- Receiving a DNA sequence through a **parameter**.
- Using **`return`** to send the position back to the caller.
- Ending the function immediately when the first match is found.

### Biological Reminder

Codons are groups of three nucleotides. In the standard genetic code, the DNA stop codons are `TAA`, `TAG`, and `TGA`.

For this exercise, assume the input is a nonempty, uppercase DNA coding-strand sequence containing only A, T, G, and C. We will examine one reading frame, beginning at index `0`.

### Task

Write a function named `find_first_stop_codon(dna_sequence)` that:

1. Accepts a DNA sequence as an input parameter.
2. Uses a `while` loop to examine complete groups of three nucleotides, starting at index `0`.
3. Uses string slicing to extract each codon.
4. Checks whether the codon is `TAA`, `TAG`, or `TGA`.
5. Returns the starting index as soon as the first stop codon is found.
6. Returns `-1` if the search finishes without finding a stop codon.

### Important Python Concepts

- **Parameter:** `dna_sequence` receives the sequence supplied when the function is called.
- **Loop:** A `while` loop moves through the sequence in steps of three.
- **Slicing:** `dna_sequence[position:position + 3]` extracts three characters beginning at `position`.
- **Membership:** `codon in stop_codons` checks whether the extracted codon belongs to the collection of stop codons.
- **Early return:** `return position` immediately ends the function and sends the position back to the caller.
- **No-match result:** `return -1` belongs after the loop, so it runs only if the search finishes without finding a match.

**Remember:** Python uses zero-based indexing. Index `0` refers to the first nucleotide, and index `3` refers to the fourth.

We check positions `0, 3, 6, 9, ...`, rather than every possible position. A final group containing fewer than three nucleotides is not a complete codon and should not be checked.

**Important:** In this function, `-1` means “no stop codon found.” It is a signal for the calling code, not a position to use when indexing the sequence.

### Test Your Function

Before running your code, predict the returned value for each sequence:

| Input sequence | What to check |
|---|---|
| `"ATGTAA"` | A stop codon after the first codon |
| `"TAAGCC"` | A stop codon at index `0` |
| `"ATGGCC"` | No stop codon |
| `"ATGTAATAG"` | More than one stop codon; return only the first |

Call your function with each sequence, store the returned value, and print it with a clear label.

**Reflection:** Why should `return -1` be placed after the loop rather than inside it?

````python

# Example: Finding the First Stop Codon in a DNA Sequence

# Define a function named find_first_stop_codon().
# The parameter 'dna_sequence' represents the DNA sequence to analyze.
def find_first_stop_codon(dna_sequence):

    # Create a list containing the three standard DNA stop codons.
    # Each stop codon consists of three nucleotides.
    # TAA, TAG, and TGA signal the end of protein translation
    # when they occur in the correct reading frame.
    stop_codons = ["TAA", "TAG", "TGA"]

    # Initialize the variable 'position' to 0.
    # Python uses zero-based indexing, so the first nucleotide is at index 0.
    # We will use this variable to track our current position in the sequence.
    position = 0

    # Use a while loop to examine the DNA sequence three nucleotides at a time.
    # len(dna_sequence) gives the total number of nucleotides.
    # Subtracting 3 ensures that at least three nucleotides remain
    # to form a complete codon.
    # The loop continues as long as 'position' is a valid codon start index.
    while position <= len(dna_sequence) - 3:

        # Extract three consecutive nucleotides using string slicing.
        # [position:position+3] starts at 'position' and stops before
        # 'position+3', giving us exactly three nucleotides.
        # For example, if position = 0, we extract indices 0, 1, and 2.
        codon = dna_sequence[position:position+3]

        # Check whether the extracted codon is in the stop_codons list.
        # The 'in' operator returns True if the codon matches
        # one of the three stop codons.
        if codon in stop_codons:

            # If a stop codon is found, return its starting index.
            # The return statement immediately ends the function,
            # so only the first stop codon is reported.
            return position

        # Move forward by three nucleotides to examine the next codon.
        # The += operator adds 3 to the current value of 'position'.
        # For example: 0 -> 3 -> 6 -> 9 -> 12, and so on.
        position += 3

    # If the loop finishes without finding any stop codon,
    # return -1 to indicate that no stop codon was found.
    # -1 is used here as a special value, not as a DNA position.
    return -1
````

**Example Usage:**

````python

# Example: Calling the Function to Find the First Stop Codon

# Create a variable named 'sequence' to store a DNA sequence.
# This sequence will be analyzed to identify the first stop codon.
sequence = "ATGCGCGATCGTAACTAGCTAGGCGTGA"

# Call the previously defined find_first_stop_codon() function.
# Pass 'sequence' as an argument to the function.
# The function examines the DNA sequence three nucleotides at a time.
# It returns the starting index of the first stop codon found.
# If no stop codon is found, the function returns -1.
# Store the returned value in the variable 'position'.
position = find_first_stop_codon(sequence)

# Use an if statement to check whether a stop codon was found.
# The != operator means "not equal to".
# If 'position' is not equal to -1, a stop codon was found.
if position != -1:

    # Display the starting index of the first stop codon.
    # The f-string inserts the value of 'position' into the message.
    print(f"Stop codon found at position {position}")

# The else block executes if the condition above is False.
# This happens when 'position' equals -1.
else:

    # Display a message indicating that no stop codon was found.
    print("No stop codon found.")
````

### Explanation:

- **Function (`find_first_stop_codon`):** This function accepts a DNA sequence as input and searches for the first occurrence of a stop codon (`TAA`, `TAG`, or `TGA`) in the reading frame starting at index `0`.

- **Stop Codons:** The list `stop_codons = ["TAA", "TAG", "TGA"]` stores the three standard DNA stop codons. The function checks whether each extracted codon matches one of these values.

- **Loop (`while`):** The `while` loop examines the DNA sequence in groups of three nucleotides (codons). The variable `position` starts at `0` and increases by `3` after each iteration, allowing the function to move from one codon to the next.

- **String Slicing (`dna_sequence[position:position+3]`):** This expression extracts three consecutive nucleotides from the DNA sequence, beginning at the current position.

- **Conditional Statement (`if`):** The function uses `if codon in stop_codons` to determine whether the current codon is a stop codon.

- **Return Values:** If a stop codon is found, the function immediately returns its starting index using `return position`. If the loop finishes without finding a stop codon, the function returns `-1`.

### Example Usage:

The script calls `find_first_stop_codon()` using the DNA sequence `"ATGCGCGATCGTAACTAGCTAGGCGTGA"` and stores the returned value in the variable `position`.

It then uses an **`if-else` statement** to determine which message to display:

- If `position != -1`, a stop codon was found, and the program prints its starting index.
- Otherwise, the program prints `"No stop codon found."`


The first stop codon in the examined reading frame is `TAG`, beginning at index `15`.

## Exercise 1: Calculating GC Content for a List of DNA Sequences

### Objective

Write a Python function that accepts a **list of DNA sequences** as input and returns a **list containing the GC content of each sequence**.

### Biological Background

As discussed earlier, **GC content** represents the percentage of guanine (G) and cytosine (C) nucleotides in a DNA sequence.

In bioinformatics, researchers often analyze many DNA sequences rather than just one. Instead of calculating GC content separately for each sequence, we can use a function and a loop to process multiple sequences automatically.

### Task

Create a function named `gc_content_list()` that:

1. Accepts a list of DNA sequences as an input parameter.
2. Creates an empty list to store the calculated GC content values.
3. Uses a `for` loop to iterate through each DNA sequence in the input list.
4. Calls the previously defined `calculate_gc_content()` function to calculate the GC percentage of each sequence.
5. Uses the `.append()` method to add each calculated percentage to the results list.
6. Returns the complete list of GC content values.

### Important Python Concepts

- **Lists:** Store multiple DNA sequences and their calculated GC percentages.
- **`for` Loop:** Iterates through each DNA sequence in the list.
- **Function Reusability:** Calls the existing `calculate_gc_content()` function instead of rewriting the calculation.
- **`.append()` Method:** Adds a new element to the end of a list.
- **Return Values:** Returns the list of calculated percentages for further analysis.

### Example

Suppose we have the following DNA sequences:

```python
sequences = ["ATGCGCGATCGTACGCTAGC", "ATATATATAT", "GGGGCCCC"]
```

The expected GC content values are:

- **First sequence:** 60.0%
- **Second sequence:** 0.0%
- **Third sequence:** 100.0%

**Expected Output:**

```text
GC Contents: [60.0, 0.0, 100.0]
```

### Key Idea

Instead of calculating GC content for each DNA sequence individually, we can **combine a loop with an existing function** to process an entire list of sequences.

This demonstrates an important principle in Python programming: **write a function once and reuse it to process multiple inputs efficiently.**

In the following code cell, we will implement this approach using a `for` loop and the `.append()` method.

````python

# Exercise 1: Calculating GC Content for a List of DNA Sequences

# Define a function named gc_content_list().
# The parameter 'sequences' represents a list of DNA sequences.
# This function will calculate the GC content of each sequence in the list.
def gc_content_list(sequences):

    # Create an empty list named 'gc_contents'.
    # The square brackets [] represent an empty list in Python.
    # We will use this list to store the GC percentage of each DNA sequence.
    gc_contents = []

    # Use a for loop to iterate through each DNA sequence in the input list.
    # During each iteration, 'seq' represents one DNA sequence.
    # The loop continues until all sequences have been processed.
    for seq in sequences:

        # Call the previously defined calculate_gc_content() function.
        # Pass the current DNA sequence ('seq') as an argument.
        # The function calculates and returns its GC percentage.
        # The .append() method adds this percentage to the end
        # of the gc_contents list.
        gc_contents.append(calculate_gc_content(seq))

    # After processing all DNA sequences, return the complete list
    # containing their calculated GC percentages.
    # The results are stored in the same order as the input sequences.
    return gc_contents


# Example Usage:

# Create a list containing three different DNA sequences.
# Each sequence is a string, and the strings are separated by commas.
sequences = ["ATGCGCGATCGTACGCTAGC", "ATATATATAT", "GGGGCCCC"]

# Call the gc_content_list() function.
# Pass the entire 'sequences' list as an argument.
# The function processes each DNA sequence and returns a list
# containing the corresponding GC percentages.
# Store the returned list in the variable 'gc_contents'.
gc_contents = gc_content_list(sequences)

# Display the calculated GC percentages for all DNA sequences.
# The print() function displays the text followed by the results list.
print("GC Contents:", gc_contents)
````

### Explanation

The `gc_content_list()` function calculates the GC content of multiple DNA sequences using a single function call.

- **Purpose:** Process a list of DNA sequences and calculate the percentage of G and C nucleotides in each sequence.
- **Processing:** A `for` loop passes each sequence to the previously defined `calculate_gc_content()` function and stores the result in a new list.
- **Output:** The function returns a list of GC percentages in the same order as the input sequences.

**Key Takeaway:** This exercise demonstrates **function reusability and list processing**. By combining an existing function with a loop, we can efficiently analyze multiple biological sequences without repeating the same calculation.

### Example Usage

For the three DNA sequences provided in the example:

- **First sequence:** Contains 60% G and C nucleotides, resulting in a GC content of **60%**.
- **Second sequence:** Contains no G or C nucleotides, resulting in a GC content of **0%**.
- **Third sequence:** Consists entirely of G and C nucleotides, resulting in a GC content of **100%**.

**Key Observation:** The returned list preserves the order of the input sequences, allowing us to match each GC content value to its corresponding DNA sequence.

## Exercise 2: Reversing a DNA Sequence Without Built-in Functions

### Objective

Write a Python script that reverses a DNA sequence using a **loop**, without using built-in reversing functions or slicing techniques such as `[::-1]`.

### Biological Context

DNA sequences are represented as strings of nucleotides (`A`, `T`, `G`, and `C`). Reversing a sequence means changing the order of its nucleotides so that the last nucleotide becomes the first.

For example:

- **Original sequence:** `ATGCCA`
- **Reversed sequence:** `ACCGTA`

**Note:** Reversing a DNA sequence is different from finding its *reverse complement*, which also requires replacing each nucleotide with its complementary base.

### Task

Use a loop to:

1. Start with an empty string to store the reversed sequence.
2. Iterate through the nucleotides in the original DNA sequence.
3. Add each nucleotide to the beginning of the new string.
4. Print the original and reversed sequences.

### Key Concept

This exercise demonstrates how **loops and string concatenation** can be used to manipulate biological sequences without relying on built-in reversing operations.

````python

# Exercise 2: Reversing a DNA Sequence Using a Loop

# Define a function that takes a DNA sequence as input
# and returns the nucleotides in reverse order.
def reverse_sequence(dna_sequence):

    # Create an empty string to store the reversed DNA sequence.
    reversed_sequence = ""

    # Use a for loop to examine each nucleotide in the original sequence.
    # During each iteration, 'nucleotide' represents one character (A, T, G, or C).
    for nucleotide in dna_sequence:

        # Add the current nucleotide to the BEGINNING of reversed_sequence.
        # The + operator joins two strings (string concatenation).
        # Placing 'nucleotide' first gradually reverses the original order.
        #
        # Example using "ATG":
        # First iteration:  "A" + ""   = "A"
        # Second iteration: "T" + "A"  = "TA"
        # Third iteration:  "G" + "TA" = "GTA"
        reversed_sequence = nucleotide + reversed_sequence

    # Return the completed reversed DNA sequence.
    return reversed_sequence


# Example Usage:

# Store the original DNA sequence in a variable.
sequence = "ATGCGCGATCGTACGCTAGC"

# Call the function and store the returned reversed sequence.
reversed_sequence = reverse_sequence(sequence)

# Display the reversed DNA sequence.
print("Reversed Sequence:", reversed_sequence)
````

### Explanation

- **Purpose:** The `reverse_sequence()` function reverses the order of nucleotides in a DNA sequence without using built-in reversing functions.

- **How It Works:** A `for` loop processes each nucleotide and adds it to the **beginning** of a new string using string concatenation (`+`). This gradually builds the sequence in reverse order.

- **Return Value:** Once all nucleotides have been processed, the function returns the reversed DNA sequence.

**Key Takeaway:** This exercise demonstrates how **loops and string concatenation** can be combined to manipulate biological sequences. It also shows that the order in which strings are joined affects the final result.

**Biological Note:** In molecular biology, a DNA palindrome typically refers to a sequence that matches its *reverse complement*, rather than simply its reverse. For this introductory programming exercise, we will use the simpler definition to practice Python string operations.

### Task

Create a function that:

2. Reverses the sequence.
3. Compares the original sequence with its reversed version.

### Key Concepts

- **String Reversal:** Changes the order of characters in a string.
- **Comparison Operator (`==`):** Checks whether two strings are identical.
- **Boolean Values:** `True` and `False` represent the result of a logical comparison.

**Key Takeaway:** This exercise demonstrates how **string manipulation, comparison operators, and Boolean return values** can be used to evaluate properties of biological sequences.

````python

# Exercise 5: Checking for Palindromic DNA Sequences

# Define a function that checks whether a DNA sequence
# reads the same forward and backward.
def is_palindromic(dna_sequence):

    # Reverse the DNA sequence using Python string slicing.
    # [::-1] means:
    #   The first ':' uses the entire string.
    #   The second ':' introduces the step value.
    #   -1 moves through the string backward, one character at a time.
    # For example: "ATTA"[::-1] produces "ATTA".
    reversed_sequence = dna_sequence[::-1]

    # Compare the original sequence with the reversed sequence.
    # The == operator checks whether two strings are identical.
    # This expression returns a Boolean value:
    # True  -> The sequence reads the same forward and backward.
    # False -> The sequence is different when reversed.
    return dna_sequence == reversed_sequence


# Example Usage:

# Define the DNA sequence to check.
sequence = "ATCGCTA"

# Call the function and store its Boolean result.
result = is_palindromic(sequence)

# Use an f-string to display whether the sequence is palindromic.
print(f"Is the sequence palindromic? {result}")
````

### Explanation

- **Purpose:** The `is_palindromic()` function checks whether a DNA sequence reads the same forward and backward.

- **String Reversal:** The slicing expression `[::-1]` reverses the order of nucleotides in the sequence.

- **Comparison and Return Value:** The equality operator (`==`) compares the original sequence with its reverse. The function returns `True` if they are identical and `False` otherwise.

### Objective

Write a Python function that takes a **DNA sequence** and a **substring** (a shorter sequence of nucleotides) as inputs and returns the number of times that substring appears in the DNA sequence.

### Biological Context

In bioinformatics, researchers often search DNA sequences for specific patterns, such as codons, short sequence motifs, or other nucleotide combinations.

For example, we may want to determine how many times the substring `"ATG"` appears in a DNA sequence. In the standard genetic code, `ATG` is a start codon, and counting its occurrences can be useful for exploring potential translation initiation sites.

### Task

Create a function that:

1. Accepts two parameters: `dna_sequence` and `substring`.
2. Searches the DNA sequence for occurrences of the specified substring.
3. Counts how many times the substring appears.
4. Returns the total count as an integer.

### Key Concepts

- **Substring:** A smaller string of consecutive characters within a larger string.
- **String Matching:** Checking whether a specific pattern appears within a sequence.
- **Counting:** Keeping track of how many times a pattern occurs.

### Example

Consider the DNA sequence:

`ATGCGATGATG`

The substring `"ATG"` appears **3 times**.

**Expected Result:** `3`

**Key Takeaway:** This exercise introduces **basic sequence pattern matching**, a fundamental technique used in bioinformatics to identify and count nucleotide patterns.

**Note:** Counting substrings is different from counting codons in a specific reading frame. A substring search can identify matches anywhere in the sequence, regardless of codon boundaries.

````python

# Exercise 6: Counting Substrings in a DNA Sequence

# Define a function that takes two parameters:
# dna_sequence: the complete DNA sequence to search.
# substring: the smaller nucleotide pattern we want to count.
def count_substring(dna_sequence, substring):

    # Initialize a counter to track the number of matches.
    count = 0

    # Start searching at index 0 (the first nucleotide).
    start = 0

    # Continue searching while enough nucleotides remain
    # to match the entire substring.
    # For example, if the substring has 3 nucleotides,
    # we need at least 3 characters remaining at each position.
    while start <= len(dna_sequence) - len(substring):

        # Extract a section of the DNA sequence using string slicing.
        # The extracted section has the same length as the substring.
        # Compare it with the target substring using ==.
        if dna_sequence[start:start+len(substring)] == substring:

            # If the two strings match, increase the counter by 1.
            count += 1

        # Move forward by ONE nucleotide to check the next position.
        # This allows the function to detect overlapping matches.
        start += 1

    # Return the total number of substring occurrences.
    return count


# Example Usage:

# Define the DNA sequence to analyze.
sequence = "ATGCGCATGCGCGATG"

# Define the nucleotide pattern to search for.
substring = "ATG"

# Call the function and store the number of matches.
count = count_substring(sequence, substring)

# Display the substring and its total number of occurrences.
print(f"The substring '{substring}' appears {count} times in the sequence.")
````

### Explanation

- **Purpose:** The `count_substring()` function counts how many times a specific nucleotide pattern, such as `"ATG"`, appears in a DNA sequence.

- **Searching the Sequence:** A `while` loop examines each possible starting position, using string slicing to extract and compare nucleotide patterns.

- **Counting Matches:** Whenever a match is found, the counter increases by `1`.

- **Return Value:** After examining all possible positions, the function returns the total number of matches.

**Key Takeaway:** Unlike the earlier stop-codon exercise, which examines nucleotides in groups of three, this function moves forward **one nucleotide at a time**. This allows it to identify patterns at any position, including overlapping occurrences.

### Understanding the Tools

- **`sequences`:** A parameter that receives the input list.
- **`for sequence in sequences`:** Takes one sequence from the list during each iteration.
- **`len(sequence)`:** Calls Python’s built-in `len()` function to determine the number of characters in that sequence.
- **`sequence_lengths.append(length)`:** Calls the list’s `.append()` method to add one result to the end of the list. It modifies the list directly.

# Using Tuples and Dictionaries with Functions

Our function will:

```python
return sequence_length, gc_percentage
```

The comma groups the results into a tuple. We can also write this explicitly as:

```python
return (sequence_length, gc_percentage)
```

### Tuple Unpacking

We can store the returned tuple in one variable:

```python
summary = summarize_sequence_tuple("ATGCGC")
```

Then, we can **unpack** its two values into separate variables:

```python
length, gc = summary
```

Python assigns values **by position**: the first value goes into `length`, and the second goes into `gc`. The variable names do not determine the assignment.

The number of variables must match the number of values for this form of unpacking.

**Before running:** Run the earlier cell defining `calculate_gc_content()`. Assume the sequence is nonempty, uppercase, and contains only A, T, G, and C.

````python
# Define the helper function first.
def calculate_gc_content(dna_sequence):
    g_count = dna_sequence.count("G")
    c_count = dna_sequence.count("C")
    return (g_count + c_count) / len(dna_sequence) * 100


# This function calls the helper and returns two results.
def summarize_sequence_tuple(sequence):
    sequence_length = len(sequence)
    gc_percentage = calculate_gc_content(sequence)

    # Return one tuple containing both values.
    return sequence_length, gc_percentage


# Call the function and store its returned tuple.
summary = summarize_sequence_tuple("ATGCGC")

print("Returned tuple:", summary)
print("Data type:", type(summary))

# Unpack the values in the order they were returned.
length, gc = summary

print("Sequence length:", length)
print(f"GC content: {gc:.2f}%")

# Practice:
# Predict the results for "AATT", then call the function to check.
````

# Error Handling and Debugging

## 1. Understanding Python Errors

### Overview

When Python cannot interpret or execute an instruction, it usually displays an error message. Learning to read that message helps us identify the problem and correct our code.

**Debugging** is the process of finding and fixing problems in a program. **Error handling** lets us respond to certain errors during execution using tools such as `try` and `except`.

### How to Read an Error Message

Start by looking for:

1. **The error type:** For example, `NameError` or `TypeError`.
2. **The explanation:** Describes what Python could not do.
3. **The indicated line:** Shows where Python encountered the problem. The underlying cause may be earlier in the code.

In Colab, errors can also occur because a required cell has not been run or the runtime has restarted.

### Common Error Types

#### a) SyntaxError

A `SyntaxError` occurs when code does not follow Python’s grammatical rules—for example, when a required colon or closing parenthesis is missing.

Python must be able to parse the code before it can execute it.

**Example:** The function definition below is missing a colon.

````python
# Intentional error: the function definition is missing a colon.
def greet(name)
    print("Hello,", name)
````

````python
# Corrected version: add the colon.
def greet(name):
    print("Hello,", name)

greet("Student")
````

### Explanation of the Previous Example

The missing colon after `def greet(name)` causes a `SyntaxError`. Python cannot execute that cell until the syntax is corrected.

#### b) TypeError

A `TypeError` occurs when an operation or function receives a type of value it does not support.

For example, `"2"` is a string, while `2` is an integer. Although they look similar, Python treats them differently.

The `+` operator can add numbers or join strings, but it cannot directly combine a string and an integer.

**Predict:** What will happen when you run the following cell?

````python
# Intentional error: combining a string and an integer.
result = "2" + 2
print(result)
````

### Fixing the Error

The correction depends on what we want to do:

- For numerical addition, use `int("2")` to convert the string into an integer.
- For joining text, use `str(2)` to convert the integer into a string.

Type conversion changes the operation Python can perform.

````python
# Numerical addition: integer + integer.
numeric_result = int("2") + 2
print("Addition:", numeric_result)  # 4

# Text concatenation: string + string.
text_result = "2" + str(2)
print("Joined text:", text_result)  # 22

# Check the types of the results.
print("Numeric result type:", type(numeric_result))
print("Text result type:", type(text_result))
````

### Explanation of the Previous Example

Python cannot directly combine a string and an integer using `+`, so it raises a `TypeError`. We must choose the appropriate conversion based on whether we want numerical addition or text concatenation.

#### c) ValueError

A `ValueError` occurs when an operation receives a supported type of input, but the particular value is unsuitable.

For example, `int()` accepts a string representing a whole number, such as `"25"`. However, `"abc"` cannot be interpreted as an integer.

**Predict:** What will happen when Python tries to convert `"abc"` into an integer?

````python
# Intentional error: this string does not represent an integer.
number = int("abc")
print(number)
````

### Understanding and Correcting the Error

Both `"abc"` and `"25"` are strings. The difference is their content: `"25"` represents a valid integer, while `"abc"` does not.

We can correct this example by supplying a valid numeric string. Later, we will use `try` and `except` to handle invalid input without stopping the program.

````python
numeric_text = "25"

# Convert a valid numeric string into an integer.
number = int(numeric_text)

print("Converted value:", number)
print("Data type:", type(number))
````

### Explanation of the Previous Example

The string `"abc"` does not represent an integer, so `int("abc")` raises a `ValueError`.

#### d) IndexError

An `IndexError` occurs when we try to access a position that does not exist in a list, tuple, string, or another sequence.

Remember that Python uses zero-based indexing. A list with three elements has nonnegative indices `0`, `1`, and `2`.

**Predict:** What will happen if we try to access index `3` in the following list?

````python
gene_names = ["BRCA1", "TP53", "EGFR"]

# Intentional error: index 3 would refer to a fourth element.
print(gene_names[3])
````

### Understanding and Correcting the Error

The list contains three elements, so index `3` is outside its range.

If we want the third element, we should use index `2`. We can also use index `-1` to access the last element of a nonempty sequence.

When debugging, check both the sequence’s length and the index you are trying to use.

````python
gene_names = ["BRCA1", "TP53", "EGFR"]

print("Number of genes:", len(gene_names))
print("Third gene:", gene_names[2])
print("Last gene:", gene_names[-1])
````

### Explanation of the Previous Example

The list `gene_names` contains three elements, with nonnegative indices `0`, `1`, and `2`. Trying to access index `3` raises an `IndexError` because there is no fourth element.

#### e) KeyError

A `KeyError` occurs when we use square brackets to retrieve a dictionary value using a key that does not exist.

Unlike lists, dictionaries use **keys** to identify their values.

**Predict:** The dictionary below contains `"sample_id"` and `"sequence"`. What will happen if we request `"gc_percentage"`?

````python
sample = {
    "sample_id": "S001",
    "sequence": "ATGCGC"
}

# Intentional error: this key does not exist in the dictionary.
print(sample["gc_percentage"])
````

### Understanding and Correcting the Error

The dictionary has no key named `"gc_percentage"`, so Python raises a `KeyError`. Key names must match exactly, including capitalization.

The appropriate response depends on what we expect:

- If the key should exist, check for a spelling mistake or a missing processing step.
- If the key is optional, check whether it exists before accessing it.

The expression `"gc_percentage" in sample` checks whether the dictionary contains that key. This avoids requesting a value that is absent, but it does not calculate the missing result.

## 2. Using `try` and `except` Blocks

### Overview

So far, we have corrected errors by changing our code or input. However, programs may receive unexpected input while running.

The `try` and `except` keywords let us handle specific exceptions and provide a helpful response instead of stopping execution.

### Syntax

```python
try:
    # Code that might raise an exception
except ExceptionType:
    # Code that handles that type of exception
```

Replace `ExceptionType` with the exception you expect, such as `ValueError`.

### How It Works

1. Python attempts to execute the instructions inside `try`.
2. If no exception occurs, Python skips the `except` block.
3. If a matching exception occurs, Python skips the remaining instructions in `try` and runs the corresponding `except` block.
4. Execution then continues after the blocks, unless another unhandled error occurs.

**Indentation matters:** Indent the instructions inside each block. Align `try` and `except` with each other.

### Example: Handle Invalid Numeric Input

We will attempt to convert text into an integer. If the text does not represent an integer, we will catch the `ValueError` and display an explanation.

**Predict:** Which messages will appear when `entered_value` is `"abc"`?

````python
entered_value = "abc"

try:
    # Attempt the conversion.
    number = int(entered_value)
    print("Converted number:", number)

except ValueError:
    # Handle text that cannot be converted into an integer.
    print("Invalid input: please provide a whole number.")

# This line runs after the error has been handled.
print("The program continues.")
````

### Explanation of the Previous Example

- The `try` block attempts to convert the input into an integer.
- If the conversion raises a `ValueError`, the matching `except` block displays a helpful message.
- After the exception is handled, the program continues.

### Handling Multiple Exceptions

A `try` block can have multiple `except` blocks to handle different types of exceptions.

In this example, the program converts a text entry into an integer and uses it to access a list element.

Two errors are possible:

- **`ValueError`:** The entry cannot be converted into an integer.
- **`IndexError`:** The integer is outside the list’s valid index range.

Python runs the first matching `except` block. Once an exception occurs, it skips the remaining instructions inside `try`; it does not resume at the failed line.

**Predict:** What will happen for each entry: `"abc"`, `"5"`, and `"1"`?

````python
gene_names = ["BRCA1", "TP53", "EGFR"]

# Change this to "5" and then "1" to test the other outcomes.
entered_index = "abc"

try:
    # This line may raise ValueError.
    index = int(entered_index)

    # This line may raise IndexError.
    selected_gene = gene_names[index]

    print("Selected gene:", selected_gene)

except ValueError:
    print("Invalid input: enter an integer index, such as 0, 1, or 2.")

except IndexError:
    print("Index out of range: this list has nonnegative indices 0–2.")

print("The program continues.")
````

### Explanation of the Previous Example

The exception depends on the input:

- `"abc"` raises a `ValueError` during conversion.
- `"5"` converts successfully but raises an `IndexError` when accessing the list.
- `"1"` selects `"TP53"` without raising an exception.

Only the matching `except` block runs.

### Optional: Using `else` and `finally`

We can add two optional blocks to a `try/except` structure:

- **`else`:** Runs if the `try` block finishes without raising an exception. Use it for actions that depend on the operation succeeding.
- **`finally`:** Runs when Python leaves the statement, whether the operation succeeded or an exception occurred. It is commonly used for cleanup, such as closing a resource.

The blocks appear in this order: `try`, `except`, `else`, `finally`.

**Predict:** Which messages will appear for `"25"`? What will change if you replace it with `"abc"`?

````python
entered_value = "25"

try:
    number = int(entered_value)

except ValueError:
    print("Invalid input: please provide a whole number.")

else:
    # Runs only when the conversion succeeds.
    print("Conversion successful:", number)

finally:
    # Runs for both inputs in this example.
    print("Conversion attempt finished.")
````

### Explanation of the Previous Example

With `"25"` as input, the conversion succeeds, so the `else` block runs. With `"abc"`, the `except ValueError` block runs instead. The `finally` block runs in both cases.

## 3. Debugging Techniques

### Overview

**Debugging** means finding and correcting problems in code. Some problems raise exceptions. Others produce incorrect results without displaying an error.

To debug effectively:

1. Identify the result you expected.
2. Compare it with the actual result.
3. Inspect the input and intermediate calculations.
4. Correct the problem and run the code again.

### a) Using Print Statements

We can place `print()` statements inside a function to inspect its input and intermediate values.

The function below is intended to calculate GC content as a percentage, but it contains a deliberate calculation mistake.

**Predict:** What percentage should `"GGCC"` produce? Run the code and use the printed values to locate the mistake.

````python
# Debugging example: this function contains an intentional mistake.
def calculate_gc_content_debug(sequence):
    g_count = sequence.count("G")
    c_count = sequence.count("C")
    total_bases = len(sequence)

    # Temporary messages help us inspect the calculation.
    print("DEBUG — input:", sequence)
    print("DEBUG — G count:", g_count)
    print("DEBUG — C count:", c_count)
    print("DEBUG — total bases:", total_bases)

    # Intentional mistake: this calculates a fraction, not a percentage.
    gc_percentage = (g_count + c_count) / total_bases

    print("DEBUG — calculated result:", gc_percentage)
    return gc_percentage


result = calculate_gc_content_debug("GGCC")
print("Reported GC percentage:", result)
````

### Investigate and Fix

The counts and sequence length are correct, but the result is `1.0` rather than the expected `100.0`.

Python does not raise an exception because the calculation is valid Python. However, it does not produce the intended percentage. This is a **logic error**.

1. Correct the calculation so it returns a percentage.
2. Run the function again with `"GGCC"`.
3. Test `"AATT"` and `"ATGC"` as well.
4. Once the results are correct, remove or comment out the temporary debugging prints.

**Remember:** Handling an exception and fixing a logic error are different tasks.

````python
# Solution: calculate a percentage rather than a fraction.
# Assumes a nonempty, uppercase sequence containing only A, T, G, and C.
def calculate_gc_content_corrected(sequence):
    g_count = sequence.count("G")
    c_count = sequence.count("C")
    total_bases = len(sequence)

    # Multiply by 100 to convert the fraction into a percentage.
    gc_percentage = (g_count + c_count) / total_bases * 100

    return gc_percentage


# Test sequences with known expected results.
test_sequences = ["GGCC", "AATT", "ATGC"]

for sequence in test_sequences:
    result = calculate_gc_content_corrected(sequence)
    print(f"Sequence: {sequence} | GC content: {result:.2f}%")
````

### Exercise: Calculate GC Content from User Input

Write a program that:

2. Removes surrounding whitespace using `.strip()` and converts letters to uppercase using `.upper()`.
3. Passes the cleaned sequence to a function.

5. Raises a `ValueError` with a helpful message if either check fails.
6. Calculates and returns the GC percentage for valid input.
7. Uses `try` and `except ValueError` around the function call to display either the result or an explanation of the invalid input.

### New Keyword: `raise`

Previously, Python raised exceptions when an operation failed. We can also deliberately raise an exception when our own validation detects a problem:

```python
raise ValueError("The sequence is empty.")
```

This stops the function and passes the exception to a matching handler, if one is available.

### Test Your Program

Run it separately with each input:

| Input | Expected behavior |
|---|---|
| `GGCC` | Display `100.00%` |
| `AATT` | Display `0.00%` |
| ` atgc ` | Clean the input and display `50.00%` |
| An empty entry | Display a helpful error message |
| `ATGX` | Display an invalid-character message |

For this exercise, we accept only the four standard DNA letters. Other symbols, including ambiguous bases such as `N`, are outside our input requirements.

Use temporary `print()` statements if you need to inspect intermediate values. Remove or comment them out after debugging.

**Reflection:** Why must we check for an empty sequence before dividing by its length?

````python
def calculate_gc_content(sequence):
    # Reject an empty sequence before dividing by its length.
    if not sequence:
        raise ValueError("The sequence is empty.")

    # Check that every character is an allowed DNA base.
    if not set(sequence).issubset({"A", "T", "G", "C"}):
        raise ValueError("The sequence contains invalid characters.")

    g_count = sequence.count("G")
    c_count = sequence.count("C")
    gc_content = (g_count + c_count) / len(sequence) * 100
    return gc_content


try:
    # Remove surrounding whitespace and standardize capitalization.
    sequence = input("Enter a DNA sequence: ").strip().upper()

    gc_content = calculate_gc_content(sequence)
    print(f"GC Content: {gc_content:.2f}%")

except ValueError as e:
    # 'e' stores the exception so we can display its message.
    print(f"Error: {e}")

finally:
    # Runs whether the input is valid or a ValueError occurs.
    print("Thank you for using the GC content calculator.")
````

- **`except ValueError as e`** catches that exception and displays its message.
- **`finally`** displays the closing message whether the calculation succeeds or a validation error occurs.

These checks address the input problems specified in this exercise. They do not guarantee that the program has no other errors.

````python
# QUESTION 3 — Define a validation function
#
# Assumptions supplied by the assignment:
# - readings is a list.
# - Each reading contains exactly two numbers.

def validate_readings(readings):
    # An empty list cannot provide a blood-pressure summary.
    if not readings:
        raise ValueError("No blood-pressure readings were submitted.")

    # Unpack each tuple into its two component values.
    for systolic, diastolic in readings:

        # Reject zero or negative values.
        if systolic <= 0 or diastolic <= 0:
            raise ValueError(
                f"Invalid reading ({systolic}, {diastolic}): "
                "both values must be greater than zero."
            )

        # Apply the assignment's simplified ordering rule.
        if systolic <= diastolic:
            raise ValueError(
                f"Invalid reading ({systolic}, {diastolic}): "
                "systolic pressure must be greater than diastolic pressure."
            )

    # This line runs only if every reading passed the checks.
    return True


# QUESTION 4 — Explain the limits of validation
#
# P003 passes these numerical checks because the list is nonempty,
# every value is positive, and systolic exceeds diastolic in each pair.
#
# However, these checks do not assess caffeine exposure, rest,
# cuff fit, device accuracy, or positioning.
#
# Numerical validity does not establish measurement accuracy.

print(
    "P003 passes numerical validation:",
    validate_readings(patients[2]["readings"])
)
````

````python
# QUESTION 5 — Summarize a list of blood-pressure readings

def summarize_bp(readings):
    # Validation raises ValueError if a record fails the checks.
    validate_readings(readings)

    systolic_values = []
    diastolic_values = []

    # Separate each paired reading into the appropriate lists.
    for systolic, diastolic in readings:
        systolic_values.append(systolic)
        diastolic_values.append(diastolic)

    # Use the actual number of readings.
    mean_systolic = sum(systolic_values) / len(systolic_values)
    mean_diastolic = sum(diastolic_values) / len(diastolic_values)

    systolic_range = max(systolic_values) - min(systolic_values)

    # Return one tuple containing three related results.
    return mean_systolic, mean_diastolic, systolic_range


# QUESTION 6 — Unpack and display P001's results

p001_summary = summarize_bp(patients[0]["readings"])

# Variable order matches the function's return order.
mean_systolic, mean_diastolic, systolic_range = p001_summary

print("Patient: P001")
print(f"Mean systolic pressure: {mean_systolic:.2f} mmHg")
print(f"Mean diastolic pressure: {mean_diastolic:.2f} mmHg")
print(f"Systolic range: {systolic_range:.2f} mmHg")

# Tuple unpacking assigns values by position, not variable name.
# Swapping variable names would mislabel the values without
# necessarily causing a Python error.
````

````python
# QUESTION 7 — Analyze one patient record

def analyze_patient(record):
    readings = record["readings"]

    # Reuse the summary function rather than repeat its calculations.
    mean_systolic, mean_diastolic, systolic_range = summarize_bp(readings)

    # Create a new dictionary without modifying the original record.
    results = {
        "patient_id": record["patient_id"],
        "reading_count": len(readings),
        "mean_systolic": mean_systolic,
        "mean_diastolic": mean_diastolic,
        "systolic_range": systolic_range,
        "measurement_note": record["measurement_note"]
    }

    return results


# QUESTION 8 — Process every record

successful_results = []
records_needing_correction = []

for record in patients:
    # Handle errors separately for each patient.
    try:
        result = analyze_patient(record)
        successful_results.append(result)

    except ValueError as e:
        patient_id = record["patient_id"]
        print(f"Record {patient_id} needs correction or collection: {e}")
        records_needing_correction.append(patient_id)


print("\nSUCCESSFUL NUMERICAL SUMMARIES")

for result in successful_results:
    print(
        f"{result['patient_id']} | "
        f"Readings: {result['reading_count']} | "
        f"Mean systolic: {result['mean_systolic']:.2f} mmHg | "
        f"Mean diastolic: {result['mean_diastolic']:.2f} mmHg | "
        f"Systolic range: {result['systolic_range']:.2f} mmHg"
    )

print("\nSuccessful records:", len(successful_results))
print("Records needing correction or collection:",
      len(records_needing_correction))
print("Affected patient IDs:", records_needing_correction)

# "Successful" means the specified numerical checks passed.
# It does not mean the blood pressure is clinically normal.
````
